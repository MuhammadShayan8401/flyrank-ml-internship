## The Question

### Research Question

Can we predict which content pages should be prioritised for review using historical search and engagement data?

### Decision

The decision is whether a content or SEO team should spend time improving a specific page.

### Action

If a page receives a high score, the team can refresh, expand, or optimise it. If the score is low, they can monitor it or focus on other pages instead.

### Unit of Analysis

One content page.

### Output

A ranked score indicating how strongly a page should be considered for review.

### Cost of a Wrong Recommendation

A false positive wastes time and resources on pages that do not improve.

A false negative means an important page is ignored, resulting in lost traffic and missed opportunities.

### Why Machine Learning?

Machine learning can analyse many features simultaneously and identify patterns that are difficult to recognise manually, making content prioritisation more efficient.

## Careful Words: What I Can and Can't Claim

This project provides **decision-support**, not guaranteed predictions. It identifies pages that appear more likely to benefit from review based on historical search and engagement data.

The results are **observational** and **directional**. They can help content teams prioritise pages for review, but they do not prove that refreshing a page will increase traffic or improve rankings.

This project cannot predict Google's algorithm or guarantee future search performance. External factors such as algorithm updates, competition, seasonality, and changes in user behaviour can all influence outcomes beyond what is captured in the dataset.
